# Inheritance

```python
#single inheritance
class School:
    def show(self):
        print(self.name)
        print(self.address)
class IIT(School):
    def display(self):
        print("Welcome to iit")
obj=IIT()
obj.name="dev"
obj.address="pune"
obj.show()
obj.display()
```

```python
#multilevel

class School:
    def show(self):
        print(self.name)
        print(self.address)
class IIT(School):
    def display(self):
        print("Welcome to iit")
class A(IIT):
    def show1(self):
        print("ground child")
obj=A()
obj.name="dev"
obj.address="pune"
obj.show()
obj.display()
obj.show1()
```

```python
#multilevel

class School:
    def show(self):
        print(self.name)
        print(self.address)
class IIT:
    def display(self):
        print("Welcome to iit")
class A(School,IIT):
    def show1(self):
        print("ground child")
obj=A()
obj.name="dev"
obj.address="pune"
obj.show()
obj.display()
obj.show1()
```
